```python
# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import os
import warnings

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler

warnings.filterwarnings("ignore")

sns.set_theme(style="whitegrid")


# ============================================================
# 2. PROJECT PATHS
# ============================================================

# This notebook is intended to be run from the project root.
DATA_DIR = os.path.join("analytics", "data")
OUTPUT_DIR = os.path.join("analytics", "outputs")

os.makedirs(DATA_DIR, exist_ok=True)
os.makedirs(OUTPUT_DIR, exist_ok=True)

LOCAL_DATA_PATH = os.path.join(
    DATA_DIR,
    "titanic.csv"
)

CLEAN_DATA_PATH = os.path.join(
    DATA_DIR,
    "titanic_clean.csv"
)


# ============================================================
# 3. LOAD DATASET
# ============================================================

print("=" * 70)
print("TASK 1 — LOAD TITANIC DATASET")
print("=" * 70)

try:
    # First try the local dataset.
    # This makes future notebook runs reproducible.
    df = pd.read_csv(LOCAL_DATA_PATH)

    print(
        f"Loaded local dataset from: {LOCAL_DATA_PATH}"
    )

except FileNotFoundError:

    # If the local file does not exist, download once
    # through Seaborn's public dataset source.
    print(
        "Local Titanic dataset not found."
    )

    print(
        "Loading Titanic dataset through Seaborn..."
    )

    df = sns.load_dataset("titanic")

    # Save a local copy for offline/reproducible use.
    df.to_csv(
        LOCAL_DATA_PATH,
        index=False
    )

    print(
        f"Local copy saved to: {LOCAL_DATA_PATH}"
    )


# Keep an untouched copy of the original dataset.
df_original = df.copy()


# ============================================================
# 4. BASIC DATASET PROFILE
# ============================================================

print("\n" + "=" * 70)
print("DATASET PROFILE")
print("=" * 70)

print(
    f"Rows    : {df.shape[0]}"
)

print(
    f"Columns : {df.shape[1]}"
)

print("\nFirst 5 rows:")
display(df.head())


print("\nDataset information:")
df.info()


print("\nNumerical summary statistics:")
display(
    df.describe()
)


print("\nCategorical summary statistics:")
display(
    df.describe(include="object")
)


# ============================================================
# 5. DUPLICATE CHECK
# ============================================================

print("\n" + "=" * 70)
print("DUPLICATE CHECK")
print("=" * 70)

duplicate_count = df.duplicated().sum()

print(
    f"Duplicate rows: {duplicate_count}"
)


# ============================================================
# 6. MISSING VALUE ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("MISSING VALUE ANALYSIS")
print("=" * 70)

missing_count = df.isnull().sum()

missing_percentage = (
    df.isnull().mean() * 100
)

missing_summary = pd.DataFrame({
    "missing_count": missing_count,
    "missing_percentage": missing_percentage.round(2)
})

missing_summary = (
    missing_summary[
        missing_summary["missing_count"] > 0
    ]
    .sort_values(
        "missing_percentage",
        ascending=False
    )
)

display(missing_summary)


# ============================================================
# 7. MISSING VALUE TREATMENT
# ============================================================

print("\n" + "=" * 70)
print("TASK 2 — MISSING VALUE TREATMENT")
print("=" * 70)

df_cleaned = df.copy()


# ------------------------------------------------------------
# DECK
# ------------------------------------------------------------
# Deck has a very high proportion of missing values.
# Therefore, it is removed rather than heavily imputed.

if "deck" in df_cleaned.columns:

    df_cleaned = df_cleaned.drop(
        columns=["deck"]
    )

    print(
        "deck: column dropped because of high missingness."
    )


# ------------------------------------------------------------
# EMBARK_TOWN
# ------------------------------------------------------------
# embark_town contains very few missing observations.
# It is redundant with embarked for the main analysis,
# so it is removed.

if "embark_town" in df_cleaned.columns:

    df_cleaned = df_cleaned.drop(
        columns=["embark_town"]
    )

    print(
        "embark_town: column dropped because it is "
        "redundant for the core analysis."
    )


# ------------------------------------------------------------
# AGE
# ------------------------------------------------------------
# Age has moderate missingness.
# Median imputation is less sensitive to extreme values
# than mean imputation.

if df_cleaned["age"].isnull().any():

    age_median = df_cleaned["age"].median()

    df_cleaned["age"] = (
        df_cleaned["age"]
        .fillna(age_median)
    )

    print(
        f"age: missing values replaced with median "
        f"({age_median:.2f})."
    )


# ------------------------------------------------------------
# EMBARKED
# ------------------------------------------------------------
# Only a very small number of observations are missing.
# These rows are removed.

if df_cleaned["embarked"].isnull().any():

    missing_embarked = (
        df_cleaned["embarked"].isnull().sum()
    )

    df_cleaned = df_cleaned.dropna(
        subset=["embarked"]
    )

    print(
        f"embarked: {missing_embarked} rows removed."
    )


# ------------------------------------------------------------
# FINAL MISSING VALUE CHECK
# ------------------------------------------------------------

print("\nRemaining missing values:")

remaining_missing = (
    df_cleaned.isnull().sum()
    .sort_values(ascending=False)
)

display(
    remaining_missing[
        remaining_missing > 0
    ]
)

print(
    "\nMissing-value treatment completed."
)


# ============================================================
# 8. FINAL DATASET SHAPE AFTER CLEANING
# ============================================================

print("\n" + "=" * 70)
print("CLEANED DATASET")
print("=" * 70)

print(
    f"Original shape : {df_original.shape}"
)

print(
    f"Cleaned shape  : {df_cleaned.shape}"
)

print(
    f"Rows removed   : "
    f"{df_original.shape[0] - df_cleaned.shape[0]}"
)


# ============================================================
# 9. SURVIVAL TARGET DISTRIBUTION
# ============================================================

print("\n" + "=" * 70)
print("SURVIVAL DISTRIBUTION")
print("=" * 70)

survival_counts = (
    df_cleaned["survived"]
    .value_counts()
    .sort_index()
)

survival_percentage = (
    df_cleaned["survived"]
    .value_counts(normalize=True)
    .sort_index()
    * 100
)

survival_summary = pd.DataFrame({
    "count": survival_counts,
    "percentage": survival_percentage.round(2)
})

display(
    survival_summary
)


# Visualization

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df_cleaned,
    x="survived"
)

plt.title(
    "Titanic Survival Distribution"
)

plt.xlabel(
    "Survived (0 = No, 1 = Yes)"
)

plt.ylabel(
    "Passenger Count"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "survival_distribution.png"
    ),
    dpi=300
)

plt.show()


# ============================================================
# 10. NUMERICAL FEATURES
# ============================================================

numerical_columns = [
    "age",
    "fare",
    "sibsp",
    "parch"
]

print("\n" + "=" * 70)
print("NUMERICAL FEATURE SUMMARY")
print("=" * 70)

display(
    df_cleaned[numerical_columns]
    .describe()
)


# ============================================================
# 11. SKEWNESS ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("SKEWNESS ANALYSIS")
print("=" * 70)

skewness_summary = pd.DataFrame({
    "feature": numerical_columns,
    "skewness": [
        df_cleaned[column].skew()
        for column in numerical_columns
    ]
})

display(
    skewness_summary.round(4)
)


# Direct fare statistics

fare_mean = df_cleaned["fare"].mean()
fare_median = df_cleaned["fare"].median()
fare_mode = df_cleaned["fare"].mode()[0]
fare_skewness = df_cleaned["fare"].skew()

print("\nFare statistics:")

print(
    f"Mean   : {fare_mean:.2f}"
)

print(
    f"Median : {fare_median:.2f}"
)

print(
    f"Mode   : {fare_mode:.2f}"
)

print(
    f"Skewness: {fare_skewness:.4f}"
)


# ============================================================
# 12. AGE DISTRIBUTION
# ============================================================

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df_cleaned,
    x="age",
    kde=True
)

plt.title(
    "Age Distribution"
)

plt.xlabel(
    "Age"
)

plt.ylabel(
    "Frequency"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "age_distribution.png"
    ),
    dpi=300
)

plt.show()


# ============================================================
# 13. FARE DISTRIBUTION
# ============================================================

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df_cleaned,
    x="fare",
    kde=True
)

plt.title(
    "Fare Distribution"
)

plt.xlabel(
    "Fare"
)

plt.ylabel(
    "Frequency"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "fare_distribution.png"
    ),
    dpi=300
)

plt.show()


# ============================================================
# 14. OUTLIER ANALYSIS — IQR METHOD
# ============================================================

print("\n" + "=" * 70)
print("TASK 3 — OUTLIER ANALYSIS USING IQR")
print("=" * 70)

outlier_results = []

for column in ["age", "fare"]:

    q1 = df_cleaned[column].quantile(0.25)

    q3 = df_cleaned[column].quantile(0.75)

    iqr = q3 - q1

    lower_bound = (
        q1 - 1.5 * iqr
    )

    upper_bound = (
        q3 + 1.5 * iqr
    )

    outlier_mask = (
        (df_cleaned[column] < lower_bound)
        |
        (df_cleaned[column] > upper_bound)
    )

    outlier_count = outlier_mask.sum()

    outlier_percentage = (
        outlier_count
        / len(df_cleaned)
        * 100
    )

    outlier_results.append({
        "feature": column,
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr,
        "lower_bound": lower_bound,
        "upper_bound": upper_bound,
        "outlier_count": outlier_count,
        "outlier_percentage": outlier_percentage
    })


outlier_summary = pd.DataFrame(
    outlier_results
)

display(
    outlier_summary.round(2)
)


# ============================================================
# 15. BOXPLOTS FOR OUTLIER VISUALIZATION
# ============================================================

for column in ["age", "fare"]:

    plt.figure(figsize=(7, 5))

    sns.boxplot(
        data=df_cleaned,
        y=column
    )

    plt.title(
        f"Boxplot of {column}"
    )

    plt.ylabel(
        column.capitalize()
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            OUTPUT_DIR,
            f"boxplot_{column}.png"
        ),
        dpi=300
    )

    plt.show()


# ============================================================
# 16. TASK 4 — SURVIVAL BY SEX
# ============================================================

print("\n" + "=" * 70)
print("SURVIVAL BY SEX")
print("=" * 70)

survival_by_sex = (
    df_cleaned
    .groupby("sex")["survived"]
    .agg(
        count="count",
        survival_rate="mean"
    )
)

survival_by_sex["survival_rate"] = (
    survival_by_sex["survival_rate"]
    * 100
)

display(
    survival_by_sex.round(2)
)


plt.figure(figsize=(7, 5))

sns.barplot(
    data=df_cleaned,
    x="sex",
    y="survived"
)

plt.title(
    "Survival Rate by Sex"
)

plt.xlabel(
    "Sex"
)

plt.ylabel(
    "Survival Rate"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "survival_by_sex.png"
    ),
    dpi=300
)

plt.show()


# ============================================================
# 17. SURVIVAL BY PASSENGER CLASS
# ============================================================

print("\n" + "=" * 70)
print("SURVIVAL BY PASSENGER CLASS")
print("=" * 70)

survival_by_class = (
    df_cleaned
    .groupby("pclass")["survived"]
    .agg(
        count="count",
        survival_rate="mean"
    )
)

survival_by_class["survival_rate"] = (
    survival_by_class["survival_rate"]
    * 100
)

display(
    survival_by_class.round(2)
)


plt.figure(figsize=(7, 5))

sns.barplot(
    data=df_cleaned,
    x="pclass",
    y="survived"
)

plt.title(
    "Survival Rate by Passenger Class"
)

plt.xlabel(
    "Passenger Class"
)

plt.ylabel(
    "Survival Rate"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "survival_by_class.png"
    ),
    dpi=300
)

plt.show()


# ============================================================
# 18. SURVIVAL BY SEX AND CLASS
# ============================================================

print("\n" + "=" * 70)
print("SURVIVAL BY SEX AND PASSENGER CLASS")
print("=" * 70)

survival_by_sex_class = (
    df_cleaned
    .groupby(
        ["sex", "pclass"]
    )["survived"]
    .mean()
    .reset_index()
)

survival_by_sex_class["survival_rate"] = (
    survival_by_sex_class["survived"]
    * 100
)

display(
    survival_by_sex_class[
        [
            "sex",
            "pclass",
            "survival_rate"
        ]
    ].round(2)
)


plt.figure(figsize=(8, 5))

sns.barplot(
    data=df_cleaned,
    x="pclass",
    y="survived",
    hue="sex"
)

plt.title(
    "Survival Rate by Passenger Class and Sex"
)

plt.xlabel(
    "Passenger Class"
)

plt.ylabel(
    "Survival Rate"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "survival_by_class_and_sex.png"
    ),
    dpi=300
)

plt.show()


# ============================================================
# 19. SURVIVAL BY EMBARKED PORT
# ============================================================

print("\n" + "=" * 70)
print("SURVIVAL BY EMBARKED PORT")
print("=" * 70)

survival_by_embarked = (
    df_cleaned
    .groupby("embarked")["survived"]
    .mean()
    .reset_index()
)

survival_by_embarked["survival_rate"] = (
    survival_by_embarked["survived"]
    * 100
)

display(
    survival_by_embarked[
        [
            "embarked",
            "survival_rate"
        ]
    ].round(2)
)


plt.figure(figsize=(7, 5))

sns.barplot(
    data=df_cleaned,
    x="embarked",
    y="survived"
)

plt.title(
    "Survival Rate by Embarkation Port"
)

plt.xlabel(
    "Embarked"
)

plt.ylabel(
    "Survival Rate"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "survival_by_embarked.png"
    ),
    dpi=300
)

plt.show()


# ============================================================
# 20. MULTIVARIATE ANALYSIS — AGE, CLASS, SURVIVAL
# ============================================================

plt.figure(figsize=(8, 6))

sns.boxplot(
    data=df_cleaned,
    x="pclass",
    y="age",
    hue="survived"
)

plt.title(
    "Age Distribution by Passenger Class and Survival"
)

plt.xlabel(
    "Passenger Class"
)

plt.ylabel(
    "Age"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "age_class_survival.png"
    ),
    dpi=300
)

plt.show()


# ============================================================
# 21. MULTIVARIATE ANALYSIS — AGE, FARE, SURVIVAL
# ============================================================

plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=df_cleaned,
    x="age",
    y="fare",
    hue="survived",
    size="pclass",
    alpha=0.7
)

plt.title(
    "Age vs Fare by Survival and Passenger Class"
)

plt.xlabel(
    "Age"
)

plt.ylabel(
    "Fare"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "age_fare_survival.png"
    ),
    dpi=300
)

plt.show()


# ============================================================
# 22. TASK 4 — REQUIRED 6-COLUMN CORRELATION MATRIX
# ============================================================

print("\n" + "=" * 70)
print("CORRELATION MATRIX")
print("=" * 70)

# Required six numerical columns.
# adult_male and alone are deliberately excluded.

numeric_columns = [
    "survived",
    "pclass",
    "age",
    "sibsp",
    "parch",
    "fare"
]

correlation_matrix = (
    df_cleaned[numeric_columns]
    .corr()
)

display(
    correlation_matrix.round(3)
)


# ============================================================
# 23. CORRELATION HEATMAP
# ============================================================

plt.figure(figsize=(9, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title(
    "Correlation Matrix — Required Six Numeric Variables"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "correlation_matrix.png"
    ),
    dpi=300
)

plt.show()


# ============================================================
# 24. STRONGEST CORRELATIONS WITH SURVIVAL
# ============================================================

print("\n" + "=" * 70)
print("CORRELATIONS WITH SURVIVAL")
print("=" * 70)

survival_correlations = (
    correlation_matrix["survived"]
    .drop("survived")
    .sort_values(
        key=lambda x: x.abs(),
        ascending=False
    )
)

display(
    survival_correlations.round(3)
)


# ============================================================
# 25. TASK 6 — Z-SCORE STANDARDIZATION CHECK
# ============================================================

print("\n" + "=" * 70)
print("TASK 6 — Z-SCORE STANDARDIZATION")
print("=" * 70)

scaler = StandardScaler()

scaled_values = scaler.fit_transform(
    df_cleaned[
        ["age", "fare"]
    ]
)

df_z = df_cleaned.copy()

df_z[
    ["age_z", "fare_z"]
] = scaled_values


# IMPORTANT:
# StandardScaler uses population standard deviation (ddof=0).
# Therefore the sanity check also uses ddof=0.

age_z_mean = df_z["age_z"].mean()
age_z_std = df_z["age_z"].std(ddof=0)

fare_z_mean = df_z["fare_z"].mean()
fare_z_std = df_z["fare_z"].std(ddof=0)


print(
    f"Age Z-score  -> Mean: {age_z_mean:.4f}, "
    f"Std: {age_z_std:.4f}"
)

print(
    f"Fare Z-score -> Mean: {fare_z_mean:.4f}, "
    f"Std: {fare_z_std:.4f}"
)


# Programmatic validation

assert np.isclose(
    age_z_mean,
    0,
    atol=1e-10
)

assert np.isclose(
    age_z_std,
    1,
    atol=1e-10
)

assert np.isclose(
    fare_z_mean,
    0,
    atol=1e-10
)

assert np.isclose(
    fare_z_std,
    1,
    atol=1e-10
)

print(
    "\nPASS: Standardization produced "
    "approximately mean = 0 and std = 1."
)


# ============================================================
# 26. SAVE CLEANED DATASET
# ============================================================

df_cleaned.to_csv(
    CLEAN_DATA_PATH,
    index=False
)

print(
    f"\nCleaned dataset saved to:"
    f"\n{CLEAN_DATA_PATH}"
)


# ============================================================
# 27. SAVE STANDARDIZED DATASET FOR REFERENCE
# ============================================================

STANDARDIZED_DATA_PATH = os.path.join(
    DATA_DIR,
    "titanic_standardized_eda.csv"
)

df_z.to_csv(
    STANDARDIZED_DATA_PATH,
    index=False
)

print(
    f"Standardized EDA dataset saved to:"
    f"\n{STANDARDIZED_DATA_PATH}"
)


# ============================================================
# 28. FINAL VALIDATION
# ============================================================

print("\n" + "=" * 70)
print("FINAL EDA VALIDATION")
print("=" * 70)

print(
    f"Original rows          : {len(df_original)}"
)

print(
    f"Cleaned rows           : {len(df_cleaned)}"
)

print(
    f"Cleaned columns        : {len(df_cleaned.columns)}"
)

print(
    f"Remaining missing cells: "
    f"{df_cleaned.isnull().sum().sum()}"
)

print(
    f"Duplicate rows         : "
    f"{df_cleaned.duplicated().sum()}"
)

print(
    f"Correlation variables  : "
    f"{len(numeric_columns)}"
)

print(
    f"Output directory       : "
    f"{OUTPUT_DIR}"
)


# ============================================================
# 29. FINAL MESSAGE
# ============================================================

print("\n" + "=" * 70)
print("MODULE 2 — EDA COMPLETED SUCCESSFULLY")
print("=" * 70)

print(
    """
Completed:

1. Dataset loading
2. Offline/local dataset fallback
3. Dataset profiling
4. Data-type inspection
5. Duplicate analysis
6. Missing-value analysis
7. Missing-value treatment
8. Numerical summary
9. Skewness analysis
10. IQR outlier analysis
11. Outlier visualizations
12. Age and fare distributions
13. Survival distribution
14. Survival by sex
15. Survival by passenger class
16. Survival by sex and class
17. Survival by embarkation port
18. Multivariate analysis
19. Required six-column correlation matrix
20. Correlation heatmap
21. Z-score standardization check
22. Cleaned dataset export
23. Standardized EDA dataset export
"""
)

print("=" * 70)
```
