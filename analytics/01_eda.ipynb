```python
# ============================================================
# ZEPTO DATA & AI PLATFORM
# MODULE 2 - TITANIC EDA
# FILE: 01_eda.ipynb
# ============================================================

import os
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler

# ============================================================
# 0. PATH SETUP
# ============================================================

# The notebook should be opened from the analytics folder.
# Therefore all files are saved directly in this folder.

BASE_DIR = os.getcwd()

print("=" * 70)
print("WORKING DIRECTORY")
print("=" * 70)
print(BASE_DIR)

# Create outputs folder directly inside the current folder
OUTPUT_DIR = os.path.join(BASE_DIR, "outputs")
os.makedirs(OUTPUT_DIR, exist_ok=True)

DATA_FILE = os.path.join(BASE_DIR, "titanic.csv")

print("\nData file:")
print(DATA_FILE)

print("\nOutput folder:")
print(OUTPUT_DIR)


# ============================================================
# TASK 1: LOAD TITANIC DATA + OFFLINE FALLBACK
# ============================================================

print("\n" + "=" * 70)
print("TASK 1: LOAD TITANIC DATA")
print("=" * 70)

try:
    df = sns.load_dataset("titanic")
    print("Titanic dataset loaded from Seaborn.")

except Exception:
    if os.path.exists(DATA_FILE):
        df = pd.read_csv(DATA_FILE)
        print("Seaborn loading failed.")
        print("Loaded Titanic dataset from local titanic.csv.")
    else:
        raise FileNotFoundError(
            "Could not load Titanic dataset and titanic.csv "
            "was not found."
        )

# Save offline fallback
df.to_csv(DATA_FILE, index=False)

print("\nOffline fallback saved successfully.")

print("\nShape:")
print(df.shape)

print("\nFirst 5 rows:")
display(df.head())

print("\nDataset information:")
df.info()

print("\nSummary statistics:")
display(df.describe())


# ============================================================
# TASK 2: MISSING VALUE PERCENTAGES
# ============================================================

print("\n" + "=" * 70)
print("TASK 2: MISSING VALUE ANALYSIS")
print("=" * 70)

missing_count = df.isnull().sum()

missing_pct = (
    df.isnull().mean() * 100
).round(2)

missing_report = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_pct
})

missing_report = missing_report[
    missing_report["Missing Count"] > 0
].sort_values(
    "Missing Percentage",
    ascending=False
)

print("\nMissing-value report:")
display(missing_report)


# ============================================================
# TASK 2: MISSING VALUE STRATEGY
# ============================================================

print("\nMissing-value strategy:")

print("deck          -> More than 30% missing -> DROP")
print("age           -> Between 5% and 30% -> MEDIAN IMPUTATION")
print("embarked      -> Less than 5% -> DROP ROWS")
print("embark_town   -> Dropped because it is redundant with embarked")

df_cleaned = df.drop(
    columns=["deck", "embark_town"]
).copy()

# Median imputation
df_cleaned["age"] = df_cleaned["age"].fillna(
    df_cleaned["age"].median()
)

# Drop rows where embarked is missing
df_cleaned = df_cleaned.dropna(
    subset=["embarked"]
).copy()

print("\nOriginal shape :", df.shape)
print("Cleaned shape  :", df_cleaned.shape)

print("\nRemaining missing values:")

remaining_missing = df_cleaned.isnull().sum()

remaining_missing = remaining_missing[
    remaining_missing > 0
]

display(remaining_missing)


# ============================================================
# SAVE CLEANED DATA
# ============================================================

CLEANED_FILE = os.path.join(
    OUTPUT_DIR,
    "titanic_cleaned.csv"
)

df_cleaned.to_csv(
    CLEANED_FILE,
    index=False
)

print("\nCleaned dataset saved to:")
print(CLEANED_FILE)


# ============================================================
# TASK 3: AGE DISTRIBUTION
# ============================================================

print("\n" + "=" * 70)
print("TASK 3: AGE DISTRIBUTION")
print("=" * 70)

plt.figure(figsize=(8, 5))

sns.histplot(
    df_cleaned["age"],
    bins=30,
    kde=True
)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "age_distribution.png"
    ),
    dpi=150,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# FARE DISTRIBUTION
# ============================================================

plt.figure(figsize=(8, 5))

sns.histplot(
    df_cleaned["fare"],
    bins=30,
    kde=True
)

plt.title("Fare Distribution")
plt.xlabel("Fare")
plt.ylabel("Frequency")

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "fare_distribution.png"
    ),
    dpi=150,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# TASK 3: IQR OUTLIER FUNCTION
# ============================================================

def calculate_iqr_outliers(series):

    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)

    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outliers = series[
        (series < lower_bound) |
        (series > upper_bound)
    ]

    return {
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": len(outliers)
    }


# ============================================================
# AGE BOXPLOT + OUTLIERS
# ============================================================

print("\n" + "=" * 70)
print("AGE OUTLIERS")
print("=" * 70)

age_outliers = calculate_iqr_outliers(
    df_cleaned["age"]
)

print(age_outliers)

plt.figure(figsize=(8, 4))

sns.boxplot(
    x=df_cleaned["age"]
)

plt.title("Age Boxplot")
plt.xlabel("Age")

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "age_boxplot.png"
    ),
    dpi=150,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# FARE BOXPLOT + OUTLIERS
# ============================================================

print("\n" + "=" * 70)
print("FARE OUTLIERS")
print("=" * 70)

fare_outliers = calculate_iqr_outliers(
    df_cleaned["fare"]
)

print(fare_outliers)

plt.figure(figsize=(8, 4))

sns.boxplot(
    x=df_cleaned["fare"]
)

plt.title("Fare Boxplot")
plt.xlabel("Fare")

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "fare_boxplot.png"
    ),
    dpi=150,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# FARE STATISTICS
# ============================================================

print("\n" + "=" * 70)
print("FARE STATISTICS")
print("=" * 70)

fare_mean = df_cleaned["fare"].mean()
fare_median = df_cleaned["fare"].median()
fare_mode = df_cleaned["fare"].mode().iloc[0]
fare_skew = df_cleaned["fare"].skew()

print(f"Fare Mean   : {fare_mean:.2f}")
print(f"Fare Median : {fare_median:.2f}")
print(f"Fare Mode   : {fare_mode:.2f}")
print(f"Fare Skew   : {fare_skew:.4f}")

if fare_skew > 0:
    print(
        "Interpretation: Fare distribution is positively/right skewed."
    )
elif fare_skew < 0:
    print(
        "Interpretation: Fare distribution is negatively/left skewed."
    )
else:
    print(
        "Interpretation: Fare distribution is approximately symmetric."
    )


# ============================================================
# TASK 4: SURVIVAL RATE BY SEX
# ============================================================

print("\n" + "=" * 70)
print("SURVIVAL RATE BY SEX")
print("=" * 70)

surv_sex = (
    df_cleaned
    .groupby("sex")["survived"]
    .mean()
)

print(surv_sex)

plt.figure(figsize=(7, 5))

sns.barplot(
    data=df_cleaned,
    x="sex",
    y="survived",
    estimator="mean"
)

plt.title("Survival Rate by Sex")
plt.xlabel("Sex")
plt.ylabel("Survival Rate")
plt.ylim(0, 1)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "survival_by_sex.png"
    ),
    dpi=150,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# SURVIVAL RATE BY PCLASS
# ============================================================

print("\n" + "=" * 70)
print("SURVIVAL RATE BY PASSENGER CLASS")
print("=" * 70)

surv_pclass = (
    df_cleaned
    .groupby("pclass")["survived"]
    .mean()
)

print(surv_pclass)

plt.figure(figsize=(7, 5))

sns.barplot(
    data=df_cleaned,
    x="pclass",
    y="survived",
    estimator="mean"
)

plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")
plt.ylim(0, 1)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "survival_by_class.png"
    ),
    dpi=150,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# SURVIVAL RATE BY SEX + PCLASS
# ============================================================

print("\n" + "=" * 70)
print("SURVIVAL RATE BY SEX AND PASSENGER CLASS")
print("=" * 70)

surv_sex_pclass = (
    df_cleaned
    .groupby(
        ["sex", "pclass"]
    )["survived"]
    .mean()
)

print(surv_sex_pclass)

plt.figure(figsize=(8, 5))

sns.barplot(
    data=df_cleaned,
    x="pclass",
    y="survived",
    hue="sex",
    estimator="mean"
)

plt.title(
    "Survival Rate by Sex and Passenger Class"
)

plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")
plt.ylim(0, 1)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "survival_by_class_and_sex.png"
    ),
    dpi=150,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# SURVIVAL RATE BY EMBARKED
# ============================================================

print("\n" + "=" * 70)
print("SURVIVAL RATE BY EMBARKED PORT")
print("=" * 70)

surv_embarked = (
    df_cleaned
    .groupby("embarked")["survived"]
    .mean()
)

print(surv_embarked)

plt.figure(figsize=(7, 5))

sns.barplot(
    data=df_cleaned,
    x="embarked",
    y="survived",
    estimator="mean"
)

plt.title("Survival Rate by Embarked Port")
plt.xlabel("Embarked Port")
plt.ylabel("Survival Rate")
plt.ylim(0, 1)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "survival_by_embarked.png"
    ),
    dpi=150,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# TASK 4: REQUIRED CORRELATION MATRIX
# ============================================================

print("\n" + "=" * 70)
print("REQUIRED CORRELATION MATRIX")
print("=" * 70)

numeric_cols = [
    "survived",
    "pclass",
    "age",
    "sibsp",
    "parch",
    "fare"
]

corr_matrix = df_cleaned[
    numeric_cols
].corr()

print("\nCorrelation matrix:")
display(corr_matrix)

plt.figure(figsize=(8, 6))

sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title(
    "Correlation Matrix - Titanic"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "correlation_matrix.png"
    ),
    dpi=150,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# TOP 2 ABSOLUTE OFF-DIAGONAL CORRELATIONS
# ============================================================

print("\n" + "=" * 70)
print("TOP 2 ABSOLUTE OFF-DIAGONAL CORRELATIONS")
print("=" * 70)

# .copy() is important because some pandas versions
# return a read-only NumPy array.
corr_abs = corr_matrix.abs().copy()

# Safely remove diagonal values
for column in corr_abs.columns:
    corr_abs.loc[column, column] = np.nan

# Convert matrix to pairs
corr_pairs = (
    corr_abs
    .stack()
    .sort_values(
        ascending=False
    )
)

seen = set()
top_two = []

for (col1, col2), absolute_value in corr_pairs.items():

    pair = tuple(
        sorted([col1, col2])
    )

    if pair not in seen:

        seen.add(pair)

        actual_value = corr_matrix.loc[
            col1,
            col2
        ]

        top_two.append(
            (
                col1,
                col2,
                actual_value,
                absolute_value
            )
        )

    if len(top_two) == 2:
        break


print("\nTop 2 correlations:")

for i, (
    col1,
    col2,
    actual_value,
    absolute_value
) in enumerate(
    top_two,
    start=1
):

    print(
        f"{i}. {col1} vs {col2} "
        f"= {actual_value:.4f} "
        f"(absolute = {absolute_value:.4f})"
    )


# ============================================================
# TASK 5: MULTIVARIATE CHART 1
# ============================================================

print("\n" + "=" * 70)
print("MULTIVARIATE CHART 1")
print("=" * 70)

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df_cleaned,
    x="age",
    y="fare",
    hue="survived",
    style="sex",
    alpha=0.7
)

plt.title(
    "Age vs Fare by Survival and Sex"
)

plt.xlabel("Age")
plt.ylabel("Fare")

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "multivariate_age_fare_survival_sex.png"
    ),
    dpi=150,
    bbox_inches="tight"
)

plt.show()

print(
    "Interpretation: This chart compares age and fare "
    "while separating passengers by survival and sex."
)

print(
    "It helps identify whether survival patterns differ "
    "across age and fare levels."
)


# ============================================================
# TASK 5: MULTIVARIATE CHART 2
# ============================================================

print("\n" + "=" * 70)
print("MULTIVARIATE CHART 2")
print("=" * 70)

plt.figure(figsize=(9, 6))

sns.boxplot(
    data=df_cleaned,
    x="pclass",
    y="age",
    hue="survived"
)

plt.title(
    "Age Distribution by Passenger Class and Survival"
)

plt.xlabel("Passenger Class")
plt.ylabel("Age")

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "multivariate_age_class_survival.png"
    ),
    dpi=150,
    bbox_inches="tight"
)

plt.show()

print(
    "Interpretation: The boxplots compare age distributions "
    "across passenger classes and survival groups."
)

print(
    "Differences in medians and spread show how age "
    "varies across these groups."
)


# ============================================================
# TASK 5: MULTIVARIATE CHART 3
# ============================================================

print("\n" + "=" * 70)
print("MULTIVARIATE CHART 3")
print("=" * 70)

plt.figure(figsize=(9, 6))

sns.barplot(
    data=df_cleaned,
    x="pclass",
    y="survived",
    hue="sex",
    estimator="mean"
)

plt.title(
    "Survival Rate by Passenger Class and Sex"
)

plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")
plt.ylim(0, 1)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "multivariate_class_sex_survival.png"
    ),
    dpi=150,
    bbox_inches="tight"
)

plt.show()

print(
    "Interpretation: This chart compares survival rates "
    "across passenger class and sex simultaneously."
)

print(
    "It shows how survival varies across combinations "
    "of class and sex."
)


# ============================================================
# TASK 5: MULTIVARIATE CHART 4
# ============================================================

print("\n" + "=" * 70)
print("MULTIVARIATE CHART 4")
print("=" * 70)

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df_cleaned,
    x="age",
    y="fare",
    hue="survived",
    size="pclass",
    sizes=(30, 250),
    alpha=0.6
)

plt.title(
    "Age vs Fare by Passenger Class and Survival"
)

plt.xlabel("Age")
plt.ylabel("Fare")

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "multivariate_age_fare_class_survival.png"
    ),
    dpi=150,
    bbox_inches="tight"
)

plt.show()

print(
    "Interpretation: This chart combines age, fare, "
    "passenger class, and survival."
)

print(
    "It gives a broader view of the relationships "
    "between these variables."
)


# ============================================================
# TASK 6: STANDARDIZATION SANITY CHECK
# ============================================================

print("\n" + "=" * 70)
print("TASK 6: STANDARDIZATION SANITY CHECK")
print("=" * 70)

scaler = StandardScaler()

scaled_values = scaler.fit_transform(
    df_cleaned[
        ["age", "fare"]
    ]
)

df_z = df_cleaned.copy()

df_z[
    ["age_z", "fare_z"]
] = scaled_values

print(
    f"Age Z-score -> "
    f"Mean: {df_z['age_z'].mean():.4f}, "
    f"Std: {df_z['age_z'].std():.4f}"
)

print(
    f"Fare Z-score -> "
    f"Mean: {df_z['fare_z'].mean():.4f}, "
    f"Std: {df_z['fare_z'].std():.4f}"
)

# Save standardized EDA data

STANDARDIZED_FILE = os.path.join(
    OUTPUT_DIR,
    "standardized_age_fare_eda.csv"
)

df_z[
    ["age", "fare", "age_z", "fare_z"]
].to_csv(
    STANDARDIZED_FILE,
    index=False
)

print(
    "\nStandardized data saved to:"
)

print(STANDARDIZED_FILE)


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("EDA COMPLETED SUCCESSFULLY")
print("=" * 70)

print("\nDataset summary:")
print(f"Original rows       : {len(df)}")
print(f"Cleaned rows        : {len(df_cleaned)}")
print(f"Original columns    : {len(df.columns)}")
print(f"Cleaned columns     : {len(df_cleaned.columns)}")

print("\nAge:")
print(
    f"IQR outliers        : "
    f"{age_outliers['Outlier Count']}"
)

print("\nFare:")
print(
    f"IQR outliers        : "
    f"{fare_outliers['Outlier Count']}"
)
print(
    f"Mean                : "
    f"{fare_mean:.2f}"
)
print(
    f"Median              : "
    f"{fare_median:.2f}"
)
print(
    f"Mode                : "
    f"{fare_mode:.2f}"
)
print(
    f"Skewness            : "
    f"{fare_skew:.4f}"
)

print("\nTop 2 correlations:")

for i, (
    col1,
    col2,
    actual_value,
    absolute_value
) in enumerate(
    top_two,
    start=1
):

    print(
        f"{i}. {col1} vs {col2}: "
        f"{actual_value:.4f}"
    )

print("\nFiles are saved inside:")
print(OUTPUT_DIR)

print("\nNo 'analytics/analytics/' path is used.")

print("\nReady for 02_modeling.ipynb.")
```
